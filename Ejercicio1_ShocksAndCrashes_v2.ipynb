Replicación parcial de Lettau y Lydvigson (2014) "Shocks and Crashes" usando un VECM estructural con la descomposición de Gonzalo-Ng (2001).

**Variables:** `c` (log consumo), `a` (log riqueza financiera), `y` (log ingreso laboral).

**Datos:** `consignas/cay_current.csv` — trimestral, 1952:Q1–2012:Q4 (muestra original de Lettau & Lydvigson).

| Inciso | Tarea | Método |
|--------|-------|--------|
| 1 | Construir `cayt` | DOLS (8 leads/lags) |
| 2 | Estimar VECM(1) | OLS ecuación por ecuación |
| 3 | Identificar shocks | Gonzalo-Ng + Cholesky |
| 4 | Interpretación económica | Análisis IRF |
| 5 | IRF en diferencias | Forma compañera VAR(2) |
| 6 | IRF en niveles | Suma acumulada |
| Bonus | Bandas de confianza | Bootstrap naïve |

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.linalg import null_space
import warnings
warnings.filterwarnings('ignore')

np.random.seed(42)
plt.rcParams.update({'figure.dpi': 100, 'font.size': 10})
```

```python
# =====================================================
# CARGAR DATOS
# =====================================================
df = pd.read_csv('consignas/cay_current.csv', parse_dates=['date'])

# Renombrar 'w' → 'a' (riqueza) para seguir la notación del examen
df = df.rename(columns={'w': 'a'})
df = df.set_index('date')

# Restringir muestra a la del paper original: 1952:Q1–2012:Q4
df = df[df.index <= '2012-12-31']

# Extraer series como arrays
c = df['c'].values          # log consumo
a = df['a'].values          # log riqueza financiera (activos)
y = df['y'].values          # log ingreso laboral
cay_lettau = df.iloc[:, -1].values  # cay pre-calculado de Lettau
dates = df.index
T = len(c)

print(f"Observaciones: {T}  (esperado ≈ 244 para 1952:Q1–2012:Q4)")
print(df.describe().round(3))

# Graficar series en niveles
fig, axes = plt.subplots(3, 1, figsize=(12, 7), sharex=True)
for ax, series, label, color in zip(axes, [c, a, y],
                                     ['c (Consumo)', 'a (Riqueza)', 'y (Ingreso)'],
                                     ['steelblue', 'darkorange', 'seagreen']):
    ax.plot(dates, series, color=color, linewidth=1.5)
    ax.set_ylabel(label)
    ax.grid(True, alpha=0.3)
ax.set_xlabel('Fecha')
fig.suptitle('Variables en Niveles (logs) — Muestra 1952:Q1–2012:Q4', fontsize=13)
plt.tight_layout()
plt.show()
```

```python
# =====================================================
# FUNCIONES AUXILIARES
# =====================================================

def estimate_dols(c, a, y, n_lags=8):
    """
    Estimación DOLS (Stock & Watson) de la relación de cointegración:

        c_t = k + alpha_a*a_t + alpha_y*y_t
              + sum_{j=1..n} [beta_a_j*Da_{t-j} + beta_y_j*Dy_{t-j}]
              + sum_{j=1..n} [delta_a_j*Da_{t+j} + delta_y_j*Dy_{t+j}] + u_t

    donde Da_t = a_t - a_{t-1}.

    Retorna:
        k, alpha_a, alpha_y  : estimadores DOLS puntuales
        cayt_full            : cay_t = c_t - alpha_a*a_t - alpha_y*y_t (muestra completa)
    """
    T = len(c)
    n = n_lags

    da = np.diff(a)
    dy = np.diff(y)

    t_start = n + 1
    t_end   = T - 1 - n
    T_eff   = t_end - t_start + 1
    t_range = np.arange(t_start, t_end + 1)

    X = np.zeros((T_eff, 3 + 4 * n))
    X[:, 0] = 1
    X[:, 1] = a[t_range]
    X[:, 2] = y[t_range]

    for j in range(1, n + 1):
        X[:, 2 + j]         = da[t_range - j - 1]
        X[:, 2 + n + j]     = dy[t_range - j - 1]
        X[:, 2 + 2*n + j]   = da[t_range + j - 1]
        X[:, 2 + 3*n + j]   = dy[t_range + j - 1]

    Y_lhs = c[t_range]

    beta, _, _, _ = np.linalg.lstsq(X, Y_lhs, rcond=None)

    k       = beta[0]
    alpha_a = beta[1]
    alpha_y = beta[2]

    cayt_full = c - alpha_a * a - alpha_y * y

    return k, alpha_a, alpha_y, cayt_full


def estimate_vecm(c, a, y, cayt, nlags=1):
    """
    Estimar VECM(1) por MCO (ecuación por ecuación):

        Dx_t = v + gamma*cay_{t-1} + Gamma*Dx_{t-1} + e_t

    Retorna: v, gamma, Gamma, Omega, resid, tstat, R2
    """
    data = np.array([c, a, y])
    T    = data.shape[1]
    n    = 3

    dx = np.diff(data, axis=1)

    Y_lhs = dx[:, 1:]

    X_rhs = np.zeros((1 + 1 + n, T - 2))
    X_rhs[0, :]  = 1
    X_rhs[1, :]  = cayt[1:T-1]
    X_rhs[2:, :] = dx[:, :T-2]

    beta = Y_lhs @ X_rhs.T @ np.linalg.inv(X_rhs @ X_rhs.T)

    v     = beta[:, 0]
    gamma = beta[:, 1]
    Gamma = beta[:, 2:]

    resid = Y_lhs - beta @ X_rhs
    T_eff = resid.shape[1]
    Omega = (resid @ resid.T) / T_eff

    k_reg   = X_rhs.shape[0]
    XXT_inv = np.linalg.inv(X_rhs @ X_rhs.T)
    tstat   = np.zeros_like(beta)
    R2      = np.zeros(n)
    for i in range(n):
        sigma2_i = np.sum(resid[i]**2) / (T_eff - k_reg)
        se_i     = np.sqrt(np.diag(sigma2_i * XXT_inv))
        tstat[i] = beta[i] / se_i
        TSS_i    = np.sum((Y_lhs[i] - Y_lhs[i].mean())**2)
        R2[i]    = 1.0 - np.sum(resid[i]**2) / TSS_i

    return v, gamma, Gamma, Omega, resid, tstat, R2


def gonzalo_ng_identification(gamma, alpha_a, alpha_y, resid, Omega):
    """
    Identificación de Gonzalo-Ng (2001):
        G = [gamma_perp' ; alpha']   (3x3)
    Retorna: G, H, H_inv, G_inv, eta
    """
    alpha = np.array([1.0, -alpha_a, -alpha_y])

    threshold    = 0.30
    gamma_norm   = np.abs(gamma) / np.linalg.norm(gamma)
    gamma_thresh = gamma.copy()
    gamma_thresh[gamma_norm < threshold] = 0.0
    gamma_perp = null_space(gamma_thresh.reshape(1, -1)).T

    for i in range(gamma_perp.shape[0]):
        dom = np.argmax(np.abs(gamma_perp[i]))
        if gamma_perp[i, dom] < 0:
            gamma_perp[i] *= -1

    G = np.vstack([gamma_perp, alpha.reshape(1, -1)])

    if np.linalg.matrix_rank(G) < 3:
        raise ValueError("La matriz G es singular — verificar supuestos de identificación.")

    G_inv   = np.linalg.inv(G)
    Sigma_u = G @ Omega @ G.T
    H       = np.linalg.cholesky(Sigma_u)
    H_inv   = np.linalg.inv(H)

    u   = G @ resid
    eta = H_inv @ u

    return G, H, H_inv, G_inv, eta


def compute_irf_vecm(Gamma, gamma, alpha_a, alpha_y, G, H, nperiods=41):
    """
    IRF de x_t (niveles) y Dx_t (diferencias) vía forma compañera VAR(2).
    irf_levels[h, var, shock], irf_diff[h, var, shock]
    """
    n     = 3
    alpha = np.array([1.0, -alpha_a, -alpha_y])

    A1 = np.eye(n) + np.outer(gamma, alpha) + Gamma
    A2 = -Gamma

    G_inv     = np.linalg.inv(G)
    A0_struct = G_inv @ H

    B = np.zeros((2*n, 2*n))
    B[:n, :n] = A1
    B[:n, n:] = A2
    B[n:, :n] = np.eye(n)

    Q_struct = np.zeros((2*n, n))
    Q_struct[:n, :] = A0_struct

    irf_levels = np.zeros((nperiods, n, n))
    B_power    = np.eye(2*n)
    for h in range(nperiods):
        impact               = B_power @ Q_struct
        irf_levels[h, :, :] = impact[:n, :]
        B_power              = B_power @ B

    irf_diff = np.zeros_like(irf_levels)
    irf_diff[0] = irf_levels[0]
    for h in range(1, nperiods):
        irf_diff[h] = irf_levels[h] - irf_levels[h-1]

    return irf_levels, irf_diff


def plot_irf(irf, var_names, shock_names, title,
             lower=None, upper=None, scale=100, ylim=None):
    """
    Graficar IRFs en grilla (var × shock). Un panel por variable×shock.
    irf[h, var, shock]
    ylim: tupla (ymin, ymax) para fijar el eje y igual en todos los paneles.
    """
    nperiods, nvars, nshocks = irf.shape
    periods = np.arange(nperiods)

    fig, axes = plt.subplots(nvars, nshocks,
                              figsize=(4*nshocks, 3.5*nvars), sharex=True)
    if nvars == 1:
        axes = axes.reshape(1, -1)

    for s in range(nshocks):
        for v in range(nvars):
            ax = axes[v, s]
            irf_vs = irf[:, v, s] * scale

            if lower is not None and upper is not None:
                ax.fill_between(periods,
                                lower[:, v, s] * scale,
                                upper[:, v, s] * scale,
                                color='lightsteelblue', alpha=0.4, label='90% IC')
                ax.plot(periods, lower[:, v, s] * scale, 'b--', linewidth=0.8)
                ax.plot(periods, upper[:, v, s] * scale, 'b--', linewidth=0.8)

            ax.plot(periods, irf_vs, color='royalblue', linewidth=2)
            ax.axhline(0, color='k', linewidth=0.6, linestyle='--', alpha=0.5)
            ax.set_title(f'Resp. de {var_names[v]}\na {shock_names[s]}', fontsize=9)
            ax.set_xlabel('Trimestres')
            ax.grid(True, alpha=0.3)
            if ylim is not None:
                ax.set_ylim(ylim)

    fig.suptitle(title, fontsize=12, y=1.01)
    plt.tight_layout()
    plt.show()


def plot_irf_byshock(irf, var_names, shock_names, title,
                     lower=None, upper=None, scale=100, ylim=None):
    """
    Graficar IRFs con un panel por shock y una línea por variable.
    irf[h, var, shock] — h=horizonte, var=variable, shock=índice.
    Colores: uno por variable (c=azul, a=naranja, y=verde).
    ylim: tupla (ymin, ymax) para fijar el eje y en todos los paneles.
    """
    nperiods, nvars, nshocks = irf.shape
    periods = np.arange(nperiods)
    colors  = ['steelblue', 'darkorange', 'seagreen']

    fig, axes = plt.subplots(1, nshocks, figsize=(5 * nshocks, 4), sharey=False)

    for s in range(nshocks):
        ax = axes[s]
        for v in range(nvars):
            ax.plot(periods, irf[:, v, s] * scale,
                    color=colors[v], linewidth=1.8, label=var_names[v])
            if lower is not None and upper is not None:
                ax.fill_between(periods,
                                lower[:, v, s] * scale,
                                upper[:, v, s] * scale,
                                color=colors[v], alpha=0.18)
        ax.axhline(0, color='k', linewidth=0.6, linestyle='--', alpha=0.5)
        ax.set_title(shock_names[s], fontsize=11)
        ax.set_xlabel('Trimestres')
        ax.legend(fontsize=8)
        ax.grid(True, alpha=0.3)
        if ylim is not None:
            ax.set_ylim(ylim)

    fig.suptitle(title, fontsize=12)
    plt.tight_layout()
    plt.show()


print("Funciones cargadas correctamente.")
```

## Inciso 1: Construcción de `cayt` por DOLS

**Método:** Dynamic OLS (Stock & Watson). Regresión de $c_t$ sobre niveles de $a_t$, $y_t$ más 8 lags y 8 leads de $\Delta a_t$ y $\Delta y_t$:

$$c_t = k + \alpha_a a_t + \alpha_y y_t + \sum_{j=1}^{8}(\beta_{a,j}\Delta a_{t-j} + \beta_{y,j}\Delta y_{t-j}) + \sum_{j=1}^{8}(\delta_{a,j}\Delta a_{t+j} + \delta_{y,j}\Delta y_{t+j}) + u_t$$

**Resultado:** $\hat{\alpha}_a$, $\hat{\alpha}_y$ y la serie $cay_t = c_t - \hat{\alpha}_a a_t - \hat{\alpha}_y y_t$.

```python
# =====================================================
# INCISO 1: DOLS
# =====================================================
k_dols, alpha_a, alpha_y, cayt_full = estimate_dols(c, a, y, n_lags=8)

print("=" * 40)
print("DOLS — Coeficientes de Cointegración")
print("=" * 40)
print(f"  Intercepto k     = {k_dols:.4f}")
print(f"  alpha_a          = {alpha_a:.4f}   (Lettau: 0.218)")
print(f"  alpha_y          = {alpha_y:.4f}   (Lettau: 0.801)")
print("\nNota: Pequeñas diferencias esperadas por la muestra extendida (2019 vs 2001).")

# Demeaning de cayt para graficar
cayt_demeaned = cayt_full - cayt_full.mean()

# Graficar: nuestro cayt vs. el de Lettau
cay_lettau_demeaned = cay_lettau - cay_lettau.mean()

fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(dates, cayt_demeaned,        color='royalblue',   linewidth=1.8,
        label='$cay_t$ (DOLS — esta replicación)')
ax.plot(dates, cay_lettau_demeaned,  color='darkorange',  linewidth=1.2,
        linestyle='--', label='$cay_t$ (Lettau — pre-calculado)')
ax.axhline(0, color='k', linewidth=0.5)
ax.set_title('Inciso 1: $cay_t = c_t - \\hat{\\alpha}_a a_t - \\hat{\\alpha}_y y_t$ (demeaned)',
             fontsize=11)
ax.set_ylabel('Desviación de la media')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Inciso 2: Estimación del VECM(1)

**Modelo en forma reducida:**

$$\Delta x_t = v + \gamma \cdot cay_{t-1} + \Gamma \cdot \Delta x_{t-1} + e_t, \quad e_t \sim (0, \Omega)$$

donde $x_t = [c_t, a_t, y_t]'$.  
Comparar con **Tabla 2** de Lettau y Lydvigson (2014).

```python
# =====================================================
# INCISO 2: ESTIMACIÓN DEL VECM
# =====================================================
v_vec, gamma, Gamma, Omega, resid, tstat, R2 = estimate_vecm(c, a, y, cayt_full)

var_names   = ['c', 'a', 'y']
shock_names = ['Perm. 1 (TFP)', 'Perm. 2 (Labor Share)', 'Transitorio']

# --- Tabla 2 al estilo Lettau & Lydvigson (2014) ---
# Columnas = ecuaciones (Δc_t, Δa_t, Δy_t)
# Filas    = regresores (cay_{t-1}, Δc_{t-1}, Δy_{t-1}, Δa_{t-1})
eq_names = ['Δc_t', 'Δa_t', 'Δy_t']
W = 14   # ancho de columna

# Mapeo: (nombre fila, coefs por ecuación, t-stats por ecuación)
# Nota: Gamma[:, 0]=Δc, Gamma[:, 1]=Δa, Gamma[:, 2]=Δy
#       Paper ordena filas Δ como: Δc, Δy, Δa → indices [0, 2, 1]
rows = [
    ('cay_{t-1}', gamma,        tstat[:, 1]),
    ('Δc_{t-1}',  Gamma[:, 0],  tstat[:, 2]),
    ('Δy_{t-1}',  Gamma[:, 2],  tstat[:, 4]),
    ('Δa_{t-1}',  Gamma[:, 1],  tstat[:, 3]),
]

SEP = "=" * (12 + W * 3)
print(SEP)
print("INCISO 2 — Tabla 2: VECM(1)  (Lettau & Lydvigson, 2014)")
print(SEP)
print(f"{'':12s}" + "".join([f"{eq:>{W}s}" for eq in eq_names]))
print("-" * (12 + W * 3))
for rname, coefs, tstats in rows:
    coef_line  = f"  {rname:10s}" + "".join([f"{b:>{W}.4f}" for b in coefs])
    tstat_line = f"  {'':10s}" + "".join([f"{f'({t:.2f})':>{W}s}" for t in tstats])
    print(coef_line)
    print(tstat_line)
print("-" * (12 + W * 3))
print(f"  {'R²':10s}" + "".join([f"{r:>{W}.2f}" for r in R2]))
print("  t-estadísticos entre paréntesis (MCO, sin corrección NW).")

print(f"\nΣ̂ (×10⁴):")
for i in range(3):
    print("  " + "  ".join([f"{Omega[i, j]*1e4:.4f}" for j in range(3)]))

print(f"\nInterceptos v (no reportados en Tabla 2 del paper):")
for i, vn in enumerate(var_names):
    print(f"  v_{vn} = {v_vec[i]:.4f}")
```

## Inciso 3: Identificación de Shocks — Descomposición Gonzalo-Ng

**Teoría (Gonzalo & Ng, 2001):**  
Construir la matriz $G = [\gamma_\perp'; \alpha']$ donde:
- $\alpha = [1, -\hat{\alpha}_a, -\hat{\alpha}_y]'$ — vector de cointegración
- $\gamma_\perp$ (2×3) — espacio nulo de $\gamma'$ (satisface $\gamma_\perp \gamma = 0$)

**Shocks antes de ortogonalizar:**
$$u_t = G e_t = \begin{bmatrix} u_t^{P_1} \\ u_t^{P_2} \\ u_t^{T} \end{bmatrix}$$

**Ortogonalización** vía Cholesky: $H = \text{chol}(G\Omega G')$, entonces $\eta_t = H^{-1} G e_t$.

```python
# =====================================================
# INCISO 3: IDENTIFICACIÓN DE GONZALO-NG
# =====================================================
G, H, H_inv, G_inv, eta = gonzalo_ng_identification(
    gamma, alpha_a, alpha_y, resid, Omega)

# Fechas para la muestra de residuos
# VECM usa t=2,...,T-1 (índice 0-based) => fechas desde índice 2
dates_resid = dates[2:]

print("=" * 45)
print("Identificación de Gonzalo-Ng")
print("=" * 45)
print(f"\nVector de cointegración α = [1, -{alpha_a:.3f}, -{alpha_y:.3f}]")
print(f"\nMatriz G (3×3):")
for row in G:
    print("  " + "  ".join([f"{x: .4f}" for x in row]))
print(f"\nrango(G) = {np.linalg.matrix_rank(G)} (debe ser 3)")
print(f"\nMatriz H (Cholesky de G·Ω·G'):")
for row in H:
    print("  " + "  ".join([f"{x: .6f}" for x in row]))
print("\nVerificación: E[η·η'] ≈ I")
Sigma_eta = (eta @ eta.T) / eta.shape[1]
for row in Sigma_eta:
    print("  " + "  ".join([f"{x: .4f}" for x in row]))
```

```python
# Normalización de signo post-Cholesky (los 3 shocks)
# La Cholesky introduce términos cruzados H[i,j] que pueden invertir η^P2 y η^T
# incluso cuando gamma_perp fue normalizado correctamente.
# Convención: P1→c LR>0, P2→a LR>0, T→a impacto>0
_irf_ref, _ = compute_irf_vecm(Gamma, gamma, alpha_a, alpha_y, G, H, nperiods=41)
_sign_refs  = [_irf_ref[40, 0, 0],   # P1: efecto LR sobre c
               _irf_ref[40, 1, 1],   # P2: efecto LR sobre a
               _irf_ref[0,  1, 2]]   # T:  efecto impacto sobre a
for _s in range(3):
    if _sign_refs[_s] < 0:
        eta[_s] *= -1

# Sumas acumuladas de shocks estructurales — comparar con Figura 5 del paper
shock_labels = ['Shock Permanente 1 (TFP)',
                'Shock Permanente 2 (Labor Share)',
                'Shock Transitorio']
colors = ['steelblue', 'darkorange', 'seagreen']

fig, axes = plt.subplots(3, 1, figsize=(14, 9), sharex=True)
for s in range(3):
    ax = axes[s]
    ax.plot(dates_resid, np.cumsum(eta[s]), color=colors[s], linewidth=1.8)
    ax.axhline(0, color='k', linewidth=0.6, linestyle='--', alpha=0.5)
    ax.set_title(shock_labels[s], fontsize=10)
    ax.set_ylabel('Suma acumulada')
    ax.grid(True, alpha=0.3)
axes[-1].set_xlabel('Fecha')

fig.suptitle('Inciso 3: Sumas Acumuladas de Shocks Estructurales\n'
             '(comparar con Figura 5 de Lettau & Lydvigson)', fontsize=11)
plt.tight_layout()
plt.show()
```

## Inciso 4: Interpretación Económica de los Shocks

Lettau y Lydvigson (2014) interpretan los dos shocks permanentes y el transitorio de la siguiente manera, basándose en las funciones impulso-respuesta (Inciso 6):

### Shock Permanente 1 → **Shock de Productividad (TFP)**

- Tiene efectos **permanentes y positivos** sobre $c_t$, $a_t$ e $y_t$ en el largo plazo.
- El efecto sobre el ingreso $y_t$ es el más grande y persistente, reflejando un aumento en la capacidad productiva de la economía.
- Afecta positivamente la riqueza financiera $a_t$ porque mejora los fundamentos de los activos.
- Es consistente con un shock de oferta agregada de largo plazo.

### Shock Permanente 2 → **Shock de Participación Laboral (Labor Share)**

- Tiene efectos **permanentes pero asimétricos** sobre las variables.
- Redistribuye entre capital y trabajo: puede aumentar la riqueza financiera mientras reduce el ingreso laboral (o viceversa).
- No afecta la productividad total de la economía pero sí la distribución del ingreso entre factores.
- Captura cambios estructurales en el mercado laboral (e.g., automatización, poder sindical).

### Shock Transitorio → **Shock de Demanda / Financiero**

- Sus efectos **se disipan en el largo plazo**: la variable $cay_t$ revierte a su media (cointegración).
- Genera desvíos temporales en la relación consumo-riqueza-ingreso.
- Lettau y Lydvigson lo asocian con crashes financieros y episodios de euforia bursátil (e.g., dot-com bubble, crisis 2008).
- La suma acumulada de este shock captura el componente transitorio de la riqueza financiera.

**Verificación formal:** Ver las IRF en niveles (Inciso 6) — los shocks permanentes generan respuestas que no vuelven a cero, el transitorio sí.

## Inciso 5: Funciones Impulso-Respuesta en Diferencias

**Definición:**
$$IR^{\Delta x}_j = E_t[\Delta x_{t+j}] - E_{t-1}[\Delta x_{t+j}]$$

Se computan vía la forma compañera del VAR(2) en niveles equivalente al VECM:
$$x_t = A_1 x_{t-1} + A_2 x_{t-2} + v + e_t$$
con $A_1 = I + \gamma\alpha' + \Gamma$, $A_2 = -\Gamma$.

$IR^{\Delta x}_h = IR^{x}_h - IR^{x}_{h-1}$

```python
# =====================================================
# INCISO 5: IRF EN DIFERENCIAS
# =====================================================
nperiods = 41
irf_levels, irf_diff = compute_irf_vecm(
    Gamma, gamma, alpha_a, alpha_y, G, H, nperiods=nperiods)

# Convención de signo consistente con Inciso 3 (P1→c LR>0, P2→a LR>0, T→a impacto>0)
_sign_refs = [irf_levels[40, 0, 0],
              irf_levels[40, 1, 1],
              irf_levels[0,  1, 2]]
for _s in range(3):
    if _sign_refs[_s] < 0:
        irf_levels[:, :, _s] *= -1
        irf_diff[:, :, _s]   *= -1

diff_var_names = ['Δc', 'Δa', 'Δy']

plot_irf_byshock(irf_diff, diff_var_names, shock_names,
                 title='Inciso 5: IRF en Diferencias — Respuesta por Shock\n'
                       '(líneas: c / a / y)',
                 scale=100)
```

## Inciso 6: Funciones Impulso-Respuesta en Niveles

**Definición:**
$$IR^{x}_j = E_t[x_{t+j}] - E_{t-1}[x_{t+j}] = \sum_{i=0}^{j} IR^{\Delta x}_i$$

Los shocks **permanentes** generan respuestas que no regresan a cero en el largo plazo.  
El shock **transitorio** genera respuestas que se disipan.

## Bonus: Bandas de Confianza Bootstrap (Naïve)

Siguiendo la metodología del curso (Blanchard-Quah class):
1. Re-samplear residuos $\hat{e}_t$ con reemplazo.
2. Regenerar series artificiales con el VECM estimado.
3. Re-estimar VECM y re-identificar shocks (Gonzalo-Ng).
4. Computar IRF para cada réplica.
5. Calcular percentiles 5% y 95% como bandas de confianza.

**Nota:** Se aplica el bootstrap a las IRF en niveles.

```python
# =====================================================
# BONUS: BOOTSTRAP NAÏVE
# =====================================================

def bootstrap_confidence_bands(c, a, y, cayt, v_vec, gamma, Gamma, resid,
                                alpha_a, alpha_y, ndraws=500, nperiods=41,
                                confidence=90):
    """
    Bootstrap naïve para impulso-respuestas del VECM.
    Re-muestrea residuos, regenera datos artificiales, re-estima VECM
    y la identificación de Gonzalo-Ng, y acumula réplicas de IRF.
    """
    print(f"Bootstrap naïve: {ndraws} réplicas, bandas de {confidence}% confianza...")
    n       = 3
    T       = len(c)
    T_eff   = resid.shape[1]
    data    = np.array([c, a, y])

    irf_levels_draws = np.zeros((nperiods, n, n, ndraws))

    for draw in range(ndraws):
        if draw % 100 == 0:
            print(f"  Réplica {draw}/{ndraws}")

        idx    = np.random.randint(0, T_eff, T_eff)
        u_boot = resid[:, idx]

        data_boot = np.zeros((n, T))
        data_boot[:, :2] = data[:, :2]

        for t in range(2, T):
            cay_lag_boot = (data_boot[0, t-1]
                            - alpha_a * data_boot[1, t-1]
                            - alpha_y * data_boot[2, t-1])
            dx_lag = data_boot[:, t-1] - data_boot[:, t-2]
            dx_t = (v_vec
                    + gamma * cay_lag_boot
                    + Gamma @ dx_lag
                    + u_boot[:, t-2])
            data_boot[:, t] = data_boot[:, t-1] + dx_t

        c_b, a_b, y_b = data_boot[0], data_boot[1], data_boot[2]
        cayt_boot = c_b - alpha_a * a_b - alpha_y * y_b

        try:
            v_b, gamma_b, Gamma_b, Omega_b, resid_b, _, _ = estimate_vecm(
                c_b, a_b, y_b, cayt_boot)
            G_b, H_b, H_inv_b, G_inv_b, eta_b = gonzalo_ng_identification(
                gamma_b, alpha_a, alpha_y, resid_b, Omega_b)
            irf_lev_b, _ = compute_irf_vecm(
                Gamma_b, gamma_b, alpha_a, alpha_y, G_b, H_b, nperiods=nperiods)
            # Convención de signo consistente con estimación principal
            _sr = [irf_lev_b[40, 0, 0], irf_lev_b[40, 1, 1], irf_lev_b[0, 1, 2]]
            for _s in range(3):
                if _sr[_s] < 0:
                    irf_lev_b[:, :, _s] *= -1
            irf_levels_draws[:, :, :, draw] = irf_lev_b
        except Exception:
            if draw > 0:
                irf_levels_draws[:, :, :, draw] = irf_levels_draws[:, :, :, draw-1]

    alpha_pct = (100 - confidence) / 2
    lower = np.percentile(irf_levels_draws, alpha_pct,       axis=3)
    upper = np.percentile(irf_levels_draws, 100 - alpha_pct, axis=3)

    return lower, upper


# Bootstrap con 1000 réplicas para resultados finales
NDRAWS = 1000

lower_boot, upper_boot = bootstrap_confidence_bands(
    c, a, y, cayt_full, v_vec, gamma, Gamma, resid,
    alpha_a, alpha_y,
    ndraws=NDRAWS, nperiods=nperiods, confidence=90
)

# Grid 3×3 (variable × shock): cada panel muestra una sola IRF + banda 90%
# Mismo rango de eje y que Inciso 6 para comparación directa
plot_irf(irf_levels, var_names, shock_names,
         title=f'Bonus: IRF en Niveles con Bandas Bootstrap Naïve (90%, {NDRAWS} réplicas)',
         lower=lower_boot, upper=upper_boot, scale=100,
         ylim=(-1.5, 2.5))
```
